```python
import cv2
import base64
import numpy as np

def display_img(base64_image: str):
    img_data = base64.b64decode(base64_image)
    nparr = np.frombuffer(img_data, np.uint8)
    img = cv2.imdecode(nparr, cv2.IMREAD_COLOR)
    cv2.imshow("Test Crosshair Placement", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

img_b64 = ""
display_img(img_b64)
```

```python
from openai import OpenAI
import time
import threading
client = OpenAI(base_url='http://localhost:8002/v1',api_key='moda')

thinking = False
def run_timer():
    global thinking
    start_time = time.time()
    while thinking:
        elapsed = time.time() - start_time
        print(f"Thinking... {elapsed:.1f}s",end='\r')
        time.sleep(0.1)
a=[
    {"role":"system","content":"You are a helpful AI assistant."},
    {"role":"user","content":"Calculate and explain what's lim(x->0) [(x-9^2)/(x-3)]"}
]
response = client.chat.completions.create(model='qwen3.5:4b',messages=a,stream=True,extra_body={"chat_template_kwargs": {"enable_thinking": True}})

thinking = False
try:
    for chunk in response:
        delta = chunk.choices[0].delta
        #print(delta)
        if hasattr(delta,'reasoning_content'):
            if delta.reasoning_content:
                if not thinking:
                    thinking = True
                    timer_thread = threading.Thread(target=run_timer, daemon=True)
                    timer_thread.start()
        elif thinking:
            thinking = False
            time.sleep(0.1)
            timer_thread.join()
        if delta.content:
            print(delta.content,end="")
except KeyboardInterrupt:
    if thinking:
        thinking = False
        timer_thread.join()
print()
```

```python
import wikipediaapi

def wiki_search(query:str, offset:int=4000) -> dict:
    #Wikipedia requires a descriptive User-Agent string to monitor traffic
    user_agent = "AIAgentPrototype_Infinity (contact: mimodassir12@gmail.com)"
    wiki = wikipediaapi.Wikipedia(
        user_agent=user_agent,
        language='en', # Forces strictly English results
        extract_format=wikipediaapi.ExtractFormat.WIKI
    )
    
    try:
        page = wiki.page(query)
    except Exception as e:
        return {
            "success": False,
            "error": f"An error occured while searching: {e}"
        }

    if not page.exists():
        return {
            "success": False,
            "error": f"No Wikipedia page found for query: '{query}'."
        }
    return {
        "success": True,
        "title": page.title,
        "url": page.fullurl,
        "summary": page.summary[:1000], # Short overview
        "full_content": page.text[:offset] # Clean body text up to your token limit
    }

result = wiki_search("Infinity")
```
